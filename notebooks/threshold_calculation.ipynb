```python
import sys
from scipy.stats import norm
```

## Movies

```python
duration_params = [
        (110, 20), (115, 18), (80, 13),  (100, 10), 
        (120, 22), (90, 18),  (100, 12), (90, 17),  
        (140, 16), (95, 13),  (120, 19), (90, 15),
        (120, 14), (110, 21), (90, 9),   (60, 8)
]
```

```python
abs_max = 0
abs_min = sys.maxsize
for mu, sigma in duration_params:
    local_min = norm.ppf(1/(10**6), loc=mu, scale=sigma)
    abs_min = min(abs_min, local_min)
    local_max = norm.ppf(1 - 1/(10**6), loc=mu, scale=sigma)
    abs_max = max(abs_max, local_max)
    
    if local_min < 1:
        print(mu, sigma)
```

```python
threshold = (abs_max + abs_min*60)/2
print(threshold)
```
